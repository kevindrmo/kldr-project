```python
# load the messing regions column
# fetch region and adminregion
import wbdata
from datetime import datetime
indicators ={
    "region":"Region Name",
    "adminregion": "Administrative Region Name"
}

df_regions_raw = wbdata.get_countries(country_id="all")

```

```python
df_regions_raw = pd.DataFrame(df_regions_raw)

df_regions_raw.head()
```

```python
def get_value_dict(cell):
    if isinstance(cell, dict):
        return cell.get("value")
    return None
df_regions_raw["region_name"] = df_regions_raw["region"].apply(get_value_dict)
df_regions_raw['admin_region_name'] = df_regions_raw['adminregion'].apply(get_value_dict)

df_regions_raw
```

```python
df_regions_raw["region_name"] = df_regions_raw["region"].apply(get_value_dict)
df_regions_raw['admin_region_name'] = df_regions_raw['adminregion'].apply(get_value_dict)
```

```python
df_regions_raw["region_final"] = df_regions_raw["region_name"].fillna(df_regions_raw["admin_region_name"])
```

```python
# filter aggregates

df_countries_only = df_regions_raw[df_regions_raw["region_final"] != "Aggregates"].copy()
df_countries_only.head()
```

```python
# selecet and rename final columns for merge

df_regions_final = df_countries_only[["id", "region_final"]].rename(
    columns = {"id": "country_code",
                "region_final": "region"}
)

df_regions_final
```

```python
df_regions_final.dropna(subset=["region"], inplace = True)

df_regions_final
```

```python
df_main = df # panel dataset

df_merged = pd.merge(
    left=df_main,
    right=df_regions_final,
    how = "left",
    on= "country_code"
)
(df_merged["region"].nunique(), df_merged["region"].unique())
```

```python
df_merged.to_csv(OUT_CLEAN/"question3"/"df_merged_for_DiD.csv")
```

### Now about the DiD analysis:

```python
import statsmodels.formula.api as smf

df_model = df_merged.copy()

df_model.dropna(subset=['Exports_Digital_Service', 'year', 'region'], inplace=True)


# a) Create the dependent variable: log_exports
df_model["log_exports"] = np.log(df_model["Exports_Digital_Service"]+1)

# b) Create a normalized time variable (starts from 0)
df_model["time"] = df_model["year"] -df_model["year"].min()


# --- 2. Define and Run the Regression Model ---

# The formula specifies the model:
# log_exports is the dependent variable.
# 'time * C(region)' tells statsmodels to include:
#   - The main effect of time ('time')
#   - The main effect of region (dummy variables for each region)
#   - The interaction effect between time and each region dummy


formula = "log_exports ~time * C(region)"

model = smf.ols(formula=formula, data = df_model)

results = model.fit()
```

```python
results.summary()
```

```python
# --- 4. Explanation of How to Interpret the Results ---

print("\n\n--- How to Interpret These Results ---")
print("This model tells us if the growth trend of digital exports differs across regions.")
print("\n1. **Baseline Region:** Statsmodels automatically chose one region as the baseline for comparison. This is the region that is *not* listed in the variable names below.")

print("\n2. **Interaction Terms (time:C(region)[...]):** These are the most important coefficients.")
print("   - A **positive** coefficient for 'time:C(region)[T.Asia]' means that, on average, digital exports in Asia grew faster each year compared to the baseline region.")
print("   - A **negative** coefficient means the growth was slower than the baseline.")
print("   - Check the 'P>|t|' column. A value less than 0.05 indicates the difference in trend is statistically significant.")

print("\n3. **Main Effects (C(region)[...]):** These coefficients show the difference in the *starting level* of digital exports in 2010 compared to the baseline region.")
```
