```python
df_raw = pd.read_csv(IN_DATASET/"final_panel_for_regression.csv")
```

```python
df_q3 = df_raw.copy()
```

```python
df_q3
```

```python
df_q3_grouped = df_q3.groupby(["year", "is_developing"])
```

```python
# For example, let's look at the data for developing countries (1) in the year 2015
group_2015_developing = df_q3_grouped.get_group((2015, 1))
group_2015_developing
```

```python
avg_adoption_df = df_q3_grouped["internet_usage_pct"].mean().reset_index()
```

```python
avg_adoption_df
```

```python
OUT_QUESTION3 = ROOT/"app"/"data"/"output"/"question3"

avg_adoption_df.to_csv(OUT_QUESTION3/"avg_adoption_long.csv", index = False)
```

#### with help of AI: for artistic display we need clean WIDE FORMAT

```python
avg_adoption_df_wide = avg_adoption_df.pivot(index = "year", 
                                            columns = "is_developing",
                                            values="internet_usage_pct"
                                            )
```
